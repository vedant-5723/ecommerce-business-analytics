Date conversion and validation

```python
date_columns = {
    "orders": "created_at",
    "order_items": "created_at",
    "order_item_refunds": "created_at",
    "products": "created_at",
    "website_sessions": "created_at",
    "website_pageviews": "created_at"
}

date_results = []

for name, column in date_columns.items():

    datasets[name][column] = pd.to_datetime(
        datasets[name][column],
        errors="coerce"
    )

    invalid_dates = datasets[name][column].isna().sum()

    date_results.append({
        "Dataset": name,
        "Date Column": column,
        "Data Type": str(datasets[name][column].dtype),
        "Invalid Dates": invalid_dates,
        "Status": "PASS" if invalid_dates == 0 else "CHECK"
    })

date_validation = pd.DataFrame(date_results)

display(date_validation)
```

Foreign Key Validation:

```python
foreign_keys = [
    ("orders", "website_session_id", "website_sessions", "website_session_id"),
    ("orders", "user_id", "website_sessions", "user_id"),
    ("order_items", "order_id", "orders", "order_id"),
    ("order_items", "product_id", "products", "product_id"),
    ("order_item_refunds", "order_item_id", "order_items", "order_item_id"),
    ("order_item_refunds", "order_id", "orders", "order_id"),
    ("website_pageviews", "website_session_id", "website_sessions", "website_session_id")
]

fk_results = []

for child_table, child_col, parent_table, parent_col in foreign_keys:

    child_values = datasets[child_table][child_col].dropna()
    parent_values = datasets[parent_table][parent_col].dropna()

    unmatched = (~child_values.isin(parent_values)).sum()

    fk_results.append({
        "Child Table": child_table,
        "Foreign Key": child_col,
        "Parent Table": parent_table,
        "Unmatched Records": unmatched,
        "Status": "PASS" if unmatched == 0 else "CHECK"
    })

fk_validation = pd.DataFrame(fk_results)

display(fk_validation)
```

Creating Clean Analytical Datasets

```python
cleaned = {
    name: df.copy()
    for name, df in datasets.items()
}

# Keep rows with valid primary keys
for name, key in primary_keys.items():
    cleaned[name] = cleaned[name].dropna(subset=[key]).copy()

# Handle missing traffic attribution fields
traffic_columns = [
    "utm_source",
    "utm_campaign",
    "utm_content",
    "http_referer"
]

for column in traffic_columns:
    cleaned["website_sessions"][column] = (
        cleaned["website_sessions"][column].fillna("Unknown")
    )

print("Cleaned dataset dimensions:")

for name, df in cleaned.items():
    print(f"{name}: {df.shape}")
```

Merging Datasets for Business Analysis:

```python
# 1. Start with order items
sales_analysis = cleaned["order_items"].copy()

# 2. Add order information
order_info = cleaned["orders"][
    ["order_id", "user_id", "created_at"]
].rename(columns={
    "created_at": "order_created_at"
})

sales_analysis = sales_analysis.merge(
    order_info,
    on="order_id",
    how="left",
    validate="many_to_one"
)

# 3. Add product information
product_info = cleaned["products"][
    ["product_id", "product_name"]
]

sales_analysis = sales_analysis.merge(
    product_info,
    on="product_id",
    how="left",
    validate="many_to_one"
)

# 4. Aggregate refunds at order-item level
refund_summary = (
    cleaned["order_item_refunds"]
    .groupby("order_item_id", as_index=False)
    ["refund_amount_usd"]
    .sum()
)

# 5. Add refund information
sales_analysis = sales_analysis.merge(
    refund_summary,
    on="order_item_id",
    how="left",
    validate="one_to_one"
)

# 6. Replace missing refund amounts with zero
sales_analysis["refund_amount_usd"] = (
    sales_analysis["refund_amount_usd"].fillna(0)
)

# 7. Create analytical columns
sales_analysis["gross_profit_usd"] = (
    sales_analysis["price_usd"] - sales_analysis["cogs_usd"]
)

sales_analysis["net_revenue_usd"] = (
    sales_analysis["price_usd"] -
    sales_analysis["refund_amount_usd"]
)

# 8. Inspect the result
print("Merged dataset shape:", sales_analysis.shape)

display(sales_analysis.head())
```

Customer Retention and Repeat Purchase Analysis

```python
# Count orders placed by each customer

customer_orders = (
    cleaned["orders"]
    .groupby("user_id")
    .agg(
        Total_Orders=("order_id", "nunique"),
        Total_Spend=("price_usd", "sum")
    )
    .reset_index()
)

# Identify repeat customers
customer_orders["Customer_Type"] = (
    customer_orders["Total_Orders"]
    .apply(lambda x: "Repeat Customer" if x > 1 else "One-Time Customer")
)

# Calculate retention-related metrics
total_customers = customer_orders["user_id"].nunique()

repeat_customers = (
    customer_orders["Total_Orders"] > 1
).sum()

one_time_customers = (
    customer_orders["Total_Orders"] == 1
).sum()

repeat_customer_percentage = (
    repeat_customers / total_customers * 100
)

one_time_customer_percentage = (
    one_time_customers / total_customers * 100
)

print("Total Customers:", total_customers)
print("Repeat Customers:", repeat_customers)
print("One-Time Customers:", one_time_customers)
print("Repeat Customer Percentage:", round(repeat_customer_percentage, 2), "%")
print("One-Time Customer Percentage:", round(one_time_customer_percentage, 2), "%")
```

Website Traffic and Conversion Analysis

```python
# Aggregate orders at website-session level
session_orders = (
    cleaned["orders"]
    .groupby("website_session_id")
    .agg(
        Total_Orders=("order_id", "nunique"),
        Revenue=("price_usd", "sum")
    )
    .reset_index()
)

# Merge order information with website sessions
traffic_analysis = cleaned["website_sessions"].merge(
    session_orders,
    on="website_session_id",
    how="left",
    validate="one_to_one"
)

# Sessions without orders have zero orders and revenue
traffic_analysis[["Total_Orders", "Revenue"]] = (
    traffic_analysis[["Total_Orders", "Revenue"]].fillna(0)
)

# Group by traffic source
source_performance = (
    traffic_analysis
    .groupby("utm_source")
    .agg(
        Total_Sessions=("website_session_id", "nunique"),
        Converted_Sessions=(
            "Total_Orders",
            lambda x: (x > 0).sum()
        ),
        Total_Revenue=("Revenue", "sum")
    )
    .reset_index()
)

# Calculate conversion rate
source_performance["Conversion_Rate_%"] = (
    source_performance["Converted_Sessions"] /
    source_performance["Total_Sessions"] * 100
)

# Sort by revenue
source_performance = source_performance.sort_values(
    "Total_Revenue",
    ascending=False
)

display(source_performance.round(2))
```

Device Performance Analysis

```python
# Aggregate orders by website session
session_orders = (
    cleaned["orders"]
    .groupby("website_session_id")
    .agg(
        Total_Orders=("order_id", "nunique"),
        Revenue=("price_usd", "sum")
    )
    .reset_index()
)

# Merge sessions with order information
device_analysis = cleaned["website_sessions"].merge(
    session_orders,
    on="website_session_id",
    how="left",
    validate="one_to_one"
)

# Fill sessions without purchases
device_analysis[["Total_Orders", "Revenue"]] = (
    device_analysis[["Total_Orders", "Revenue"]].fillna(0)
)

# Analyze device performance
device_performance = (
    device_analysis
    .groupby("device_type")
    .agg(
        Total_Sessions=("website_session_id", "nunique"),
        Converted_Sessions=(
            "Total_Orders",
            lambda x: (x > 0).sum()
        ),
        Total_Orders=("Total_Orders", "sum"),
        Total_Revenue=("Revenue", "sum")
    )
    .reset_index()
)

device_performance["Conversion_Rate_%"] = (
    device_performance["Converted_Sessions"] /
    device_performance["Total_Sessions"] * 100
)

device_performance["Revenue_Per_Session"] = (
    device_performance["Total_Revenue"] /
    device_performance["Total_Sessions"]
)

display(device_performance.round(2))
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    device_performance["device_type"],
    device_performance["Conversion_Rate_%"]
)

plt.title("Conversion Rate by Device")
plt.xlabel("Device Type")
plt.ylabel("Conversion Rate (%)")

plt.tight_layout()
plt.show()
```

```python
# Count pageviews for each website session

pageview_summary = (
    cleaned["website_pageviews"]
    .groupby("website_session_id")
    .agg(
        Total_Pageviews=("website_pageview_id", "nunique")
    )
    .reset_index()
)

# Merge pageview information with website sessions

engagement_analysis = cleaned["website_sessions"].merge(
    pageview_summary,
    on="website_session_id",
    how="left",
    validate="one_to_one"
)

# Sessions with no recorded pageviews
engagement_analysis["Total_Pageviews"] = (
    engagement_analysis["Total_Pageviews"].fillna(0)
)

# Create engagement categories
engagement_analysis["Session_Type"] = (
    engagement_analysis["Total_Pageviews"]
    .apply(
        lambda x: "Single Page Session"
        if x == 1
        else "Multiple Page Session"
        if x > 1
        else "No Pageviews"
    )
)

# Overall engagement KPIs
total_sessions = len(engagement_analysis)

single_page_sessions = (
    engagement_analysis["Total_Pageviews"] == 1
).sum()

multiple_page_sessions = (
    engagement_analysis["Total_Pageviews"] > 1
).sum()

average_pageviews = (
    engagement_analysis["Total_Pageviews"].mean()
)

print("Total Sessions:", total_sessions)
print("Single Page Sessions:", single_page_sessions)
print("Multiple Page Sessions:", multiple_page_sessions)
print("Average Pageviews per Session:", round(average_pageviews, 2))
print(
    "Single Page Session Percentage:",
    round(single_page_sessions / total_sessions * 100, 2),
    "%"
)
```

Q1. Total Revenue by Month

```python
query_1 = """
SELECT
    strftime('%Y-%m', created_at) AS month,
    ROUND(SUM(price_usd), 2) AS total_revenue
FROM orders
GROUP BY strftime('%Y-%m', created_at)
ORDER BY month;
"""

q1_result = pd.read_sql_query(query_1, conn)

display(q1_result.head(15))
```

Q2. Top Products by Revenue

```python
query_2 = """
SELECT
    p.product_id,
    p.product_name,
    ROUND(SUM(oi.price_usd), 2) AS total_revenue,
    COUNT(oi.order_item_id) AS units_sold
FROM order_items oi
JOIN products p
    ON oi.product_id = p.product_id
GROUP BY
    p.product_id,
    p.product_name
ORDER BY total_revenue DESC
LIMIT 10;
"""

q2_result = pd.read_sql_query(query_2, conn)

display(q2_result)
```

Q5. Customers with More Than One Order

```python
query_5 = """
SELECT
    user_id,
    COUNT(DISTINCT order_id) AS total_orders
FROM orders
GROUP BY user_id
HAVING COUNT(DISTINCT order_id) > 1
ORDER BY total_orders DESC, user_id;
"""

q5_result = pd.read_sql_query(query_5, conn)

print("Number of repeat customers:", len(q5_result))

display(q5_result.head(15))
```

Q6. Repeat Customer Percentage

```python
query_6 = """
SELECT
    COUNT(*) AS total_customers,

    SUM(
        CASE
            WHEN order_count > 1 THEN 1
            ELSE 0
        END
    ) AS repeat_customers,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN order_count > 1 THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS repeat_customer_percentage

FROM (
    SELECT
        user_id,
        COUNT(DISTINCT order_id) AS order_count
    FROM orders
    GROUP BY user_id
) AS customer_summary;
"""

q6_result = pd.read_sql_query(query_6, conn)

display(q6_result)
```

Q10. Monthly Sales Growth

```python
query_10 = """
WITH monthly_revenue AS (
    SELECT
        strftime('%Y-%m', created_at) AS month,
        SUM(price_usd) AS total_revenue
    FROM orders
    GROUP BY strftime('%Y-%m', created_at)
),

revenue_comparison AS (
    SELECT
        month,
        total_revenue,
        LAG(total_revenue) OVER (
            ORDER BY month
        ) AS previous_month_revenue
    FROM monthly_revenue
)

SELECT
    month,
    ROUND(total_revenue, 2) AS total_revenue,
    ROUND(previous_month_revenue, 2) AS previous_month_revenue,
    ROUND(
        (total_revenue - previous_month_revenue)
        * 100.0 / NULLIF(previous_month_revenue, 0),
        2
    ) AS monthly_growth_percentage
FROM revenue_comparison
ORDER BY month;
"""

q10_result = pd.read_sql_query(query_10, conn)

display(q10_result.head(15))
```

Q11. Website Users Who Never Ordered

Q12. Overall Business KPI Analysis

```python
query_12 = """
WITH refund_summary AS (
    SELECT
        order_item_id,
        SUM(refund_amount_usd) AS refund_amount
    FROM order_item_refunds
    GROUP BY order_item_id
),

item_metrics AS (
    SELECT
        COUNT(oi.order_item_id) AS total_items,
        COUNT(DISTINCT oi.order_id) AS item_order_count,
        SUM(oi.price_usd) AS gross_revenue,
        SUM(oi.cogs_usd) AS total_cogs,
        SUM(oi.price_usd - oi.cogs_usd) AS gross_profit,
        SUM(COALESCE(r.refund_amount, 0)) AS total_refunds,
        COUNT(DISTINCT p.product_id) AS products_sold
    FROM order_items oi
    JOIN products p
        ON oi.product_id = p.product_id
    LEFT JOIN refund_summary r
        ON oi.order_item_id = r.order_item_id
),

order_metrics AS (
    SELECT
        COUNT(DISTINCT order_id) AS total_orders,
        COUNT(DISTINCT user_id) AS total_customers,
        SUM(price_usd) AS order_revenue,
        AVG(price_usd) AS average_order_value
    FROM orders
)

SELECT
    o.total_orders,
    o.total_customers,
    i.total_items,
    i.products_sold,
    ROUND(i.gross_revenue, 2) AS gross_revenue,
    ROUND(i.total_cogs, 2) AS total_cogs,
    ROUND(i.gross_profit, 2) AS gross_profit,
    ROUND(i.total_refunds, 2) AS total_refunds,
    ROUND(i.gross_revenue - i.total_refunds, 2) AS net_revenue,
    ROUND(o.average_order_value, 2) AS average_order_value,
    ROUND(
        100.0 * i.gross_profit / NULLIF(i.gross_revenue, 0),
        2
    ) AS gross_profit_margin_percentage,
    ROUND(
        100.0 * i.total_refunds / NULLIF(i.gross_revenue, 0),
        2
    ) AS refund_percentage
FROM order_metrics o
CROSS JOIN item_metrics i;
"""

q12_result = pd.read_sql_query(query_12, conn)

display(q12_result.T)
```
